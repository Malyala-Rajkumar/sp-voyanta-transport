```python
from datetime import datetime

def extract_essential_flight_info(raw_data, max_results=5):
    flights = raw_data.get("data", [])
    compact_flights = []
    
    for f in flights[:max_results]:
        dep_time_str = f.get("departure", {}).get("scheduled")
        arr_time_str = f.get("arrival", {}).get("scheduled")
        
        # Calculate duration if timestamps are present
        duration = "N/A"
        if dep_time_str and arr_time_str:
            dep_dt = datetime.fromisoformat(dep_time_str)
            arr_dt = datetime.fromisoformat(arr_time_str)
            diff_min = int((arr_dt - dep_dt).total_seconds() / 60)
            duration = f"{diff_min // 60}h {diff_min % 60}m"

        compact_flights.append({
            "airline": f.get("airline", {}).get("name"),
            "flight_no": f.get("flight", {}).get("iata"),
            "departure": f.get("departure", {}).get("iata") + " " + (dep_time_str[11:16] if dep_time_str else ""),
            "arrival": f.get("arrival", {}).get("iata") + " " + (arr_time_str[11:16] if arr_time_str else ""),
            "duration": duration,
            "status": f.get("flight_status")
        })
        
    return compact_flights
```

```python
import difflib

# Major Indian Cities & Tourist Destinations mapped to Airport IATA Codes
INDIAN_AIRPORTS = {
    # Major Metros
    "hyderabad": {"code": "HYD", "name": "Rajiv Gandhi Intl, Hyderabad"},
    "bangalore": {"code": "BLR", "name": "Kempegowda Intl, Bengaluru"},
    "bengaluru": {"code": "BLR", "name": "Kempegowda Intl, Bengaluru"},
    "delhi": {"code": "DEL", "name": "Indira Gandhi Intl, New Delhi"},
    "mumbai": {"code": "BOM", "name": "Chhatrapati Shivaji Intl, Mumbai"},
    "chennai": {"code": "MAA", "name": "Chennai Intl, Chennai"},
    "kolkata": {"code": "CCU", "name": "Netaji Subhash Chandra Bose Intl, Kolkata"},
    
    # Kerala (Users often type "Kerala", "Kochi", or "Trivandrum")
    "kerala": {"code": "COK", "name": "Cochin International Airport (Primary Kerala Hub)"},
    "kochi": {"code": "COK", "name": "Cochin International Airport, Kochi"},
    "cochin": {"code": "COK", "name": "Cochin International Airport, Kochi"},
    "trivandrum": {"code": "TRV", "name": "Thiruvananthapuram Intl, Trivandrum"},
    "thiruvananthapuram": {"code": "TRV", "name": "Thiruvananthapuram Intl, Trivandrum"},
    "calicut": {"code": "CCJ", "name": "Calicut Intl Airport, Kozhikode"},
    "kozhikode": {"code": "CCJ", "name": "Calicut Intl Airport, Kozhikode"},
    
    # Top Holiday Spots
    "goa": {"code": "GOI", "name": "Dabolim Airport / MOPA (GOX), Goa"},
    "jaipur": {"code": "JAI", "name": "Jaipur Intl, Rajasthan"},
    "varanasi": {"code": "VNS", "name": "Lal Bahadur Shastri Intl, Varanasi"},
    "srinagar": {"code": "SXR", "name": "Sheikh ul-Alam Intl, Kashmir"},
    "chandigarh": {"code": "IXC", "name": "Shaheed Bhagat Singh Intl, Chandigarh"},
    "ahmedabad": {"code": "AMD", "name": "Sardar Vallabhbhai Patel Intl, Ahmedabad"},
    "pune": {"code": "PNQ", "name": "Pune Airport, Maharashtra"}
}

def resolve_airport_code(location_input: str) -> str:
    """Takes any city name or airport code and returns the valid 3-letter IATA code."""
    loc = location_input.strip().lower()
    
    # If already a valid 3-letter code (e.g., 'HYD')
    if len(loc) == 3 and loc.upper() in [v["code"] for v in INDIAN_AIRPORTS.values()]:
        return loc.upper()
        
    # Direct dictionary match
    if loc in INDIAN_AIRPORTS:
        return INDIAN_AIRPORTS[loc]["code"]
        
    # Fuzzy match (e.g., handles typos like 'Hyderbad' or 'Bengalore')
    matches = difflib.get_close_matches(loc, INDIAN_AIRPORTS.keys(), n=1, cutoff=0.6)
    if matches:
        matched_key = matches[0]
        print(f"-> Interpreted '{location_input}' as '{INDIAN_AIRPORTS[matched_key]['name']}' ({INDIAN_AIRPORTS[matched_key]['code']})")
        return INDIAN_AIRPORTS[matched_key]["code"]
        
    raise ValueError(f"Could not find an airport for '{location_input}'. Please check the city name.")
```

```python
import serpapi

def get_flight_itinerary(origin_city: str, destination_city: str, travel_date: str):
    # Automatically resolve city names to IATA codes
    dep_code = resolve_airport_code(origin_city)
    arr_code = resolve_airport_code(destination_city)
    
    client = serpapi.Client(api_key="62705f463d8b769d216c7cf684ecaddfb58602456ad003fc667ee72dfe2b9073")
    
    params = {
        "engine": "google_flights",
        "departure_id": dep_code,
        "arrival_id": arr_code,
        "outbound_date": travel_date,
        "type": "2",
        "currency": "INR",
        "gl": "in",
        "hl": "en"
    }
    
    results = client.search(params)
    flights = results.get("best_flights", []) or results.get("other_flights", [])
    
    clean_results = []
    for f in flights[:3]:
        legs = f.get("flights", [])
        clean_results.append({
            "airline": legs[0].get("airline"),
            "flight_no": legs[0].get("flight_number"),
            "route": f"{dep_code} -> {arr_code}",
            "departure": legs[0].get("departure_airport", {}).get("time"),
            "arrival": legs[-1].get("arrival_airport", {}).get("time"),
            "duration": f"{f.get('total_duration')} mins",
            "price": f"₹{f.get('price')}"
        })
    return clean_results

# Now the user can pass natural city names:
flights = get_flight_itinerary(origin_city="Hyderabad", destination_city="chennai", travel_date="2026-10-25")
flights
```

```python
import requests

def search_redbus_live(source="Hyderabad", destination="Bangalore", doj="2026-10-25"):
    # redBus public search endpoint
    # Note: Requires realistic browser User-Agent headers
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
        "Accept": "application/json"
    }
    
    # 1. Resolve City IDs (e.g., Hyderabad = 124, Bangalore = 122)
    # 2. Query available trips
    url = f"https://www.redbus.in/api/getAvailableTrips?fromCity={source}&toCity={destination}&doj={doj}"
    
    try:
        res = requests.get(url, headers=headers, timeout=10)
        data = res.json()
        
        buses = []
        for trip in data.get("availableTrips", [])[:5]:
            buses.append({
                "mode": "Bus",
                "operator": trip.get("travelsName"),
                "bus_type": trip.get("busType"),
                "departure": trip.get("departureTime"),
                "arrival": trip.get("arrivalTime"),
                "fare": f"₹{trip.get('fareDetails', [{}])[0].get('baseFare')}",
                "seats_available": trip.get("availableSeats")
            })
        return buses
    except Exception as e:
        return f"Error fetching redBus data: {e}"
```

```python
import requests
from datetime import datetime
from google.transit import gtfs_realtime_pb2
from google.protobuf.json_format import MessageToDict

def fetch_and_enrich_bus_data(route_id_filter=None, max_results=5):
    url = "https://otd.delhi.gov.in/api/realtime/VehiclePositions.pb?key=gh9lYzYwBXt5TKekre2CD56VUBZ78eLu"
    
    feed = gtfs_realtime_pb2.FeedMessage()
    response = requests.get(url, timeout=15)
    
    if response.status_code != 200:
        return f"Error fetching feed: HTTP {response.status_code}"
        
    feed.ParseFromString(response.content)
    current_time = datetime.now()
    
    enriched_buses = []
    
    for entity in feed.entity:
        if not entity.HasField("vehicle"):
            continue
            
        v = entity.vehicle
        route_id = v.trip.route_id
        
        # If user asked for a specific route (e.g., route 2141), filter it
        if route_id_filter and str(route_id) != str(route_id_filter):
            continue
            
        reg_no = v.vehicle.id
        
        # 1. Identify Bus Category from Number Plate
        if "EV" in reg_no:
            bus_category = "Zero-Emission Electric AC Bus"
        elif "PD" in reg_no or "PC" in reg_no:
            bus_category = "DTC Low-Floor CNG Bus"
        else:
            bus_category = "Standard City Bus"
            
        # 2. Speed Conversion (m/s to km/h)
        speed_kmh = round(v.position.speed * 3.6, 1) if v.position.speed else 0.0
        movement_status = "In Transit (Moving)" if speed_kmh > 3 else "Halted / At Bus Stop"
        
        # 3. Last GPS Heartbeat
        bus_time = datetime.fromtimestamp(v.timestamp) if v.timestamp else current_time
        seconds_ago = int((current_time - bus_time).total_seconds())
        
        # 4. Schedule Type
        sched_rel = v.trip.schedule_relationship
        sched_type = "Regular Schedule" if sched_rel == 0 else "Special / Added Service"
        
        enriched_buses.append({
            "vehicle_number": reg_no,
            "category": bus_category,
            "route_id": route_id,
            "trip_id": v.trip.trip_id,
            "trip_start_time": v.trip.start_time,
            "current_location": {
                "latitude": round(v.position.latitude, 5),
                "longitude": round(v.position.longitude, 5)
            },
            "speed": f"{speed_kmh} km/h",
            "movement_status": movement_status,
            "schedule_type": sched_type,
            "last_seen": f"{seconds_ago} seconds ago"
        })
        
        if len(enriched_buses) >= max_results:
            break
            
    return enriched_buses

# Test the extractor
buses_for_llm = fetch_and_enrich_bus_data(max_results=3)
print(buses_for_llm)
```

```python
# Install geopy if you don't have it: pip install geopy
from geopy.geocoders import Nominatim

geolocator = Nominatim(user_agent="delhi_bus_tracker")

def get_location_name(lat, lon):
    try:
        location = geolocator.reverse((lat, lon), timeout=5)
        address = location.raw.get("address", {})
        
        # Pick the most human-readable parts (Road, Suburb/Area, City)
        road = address.get("road") or address.get("suburb") or address.get("neighbourhood")
        city_area = address.get("city_district") or address.get("city") or "New Delhi"
        
        if road:
            return f"Near {road}, {city_area}"
        return location.address.split(",")[0]
    except Exception:
        return "Central Delhi area"

# Test with your bus coordinates:
area = get_location_name(28.60397, 77.22579)
print(f"Human-friendly Location: {area}")
# Output: Near Lodhi Road, New Delhi
```
