```python
import json
import serpapi

def search_google_flights(
    departure_id: str,
    arrival_id: str,
    outbound_date: str,
    return_date: str = None,
    currency: str = "INR"
):
    """
    Searches Google Flights via SerpApi and returns structured, cleaned flight data.
    """
    API_KEY = "62705f463d8b769d216c7cf684ecaddfb58602456ad003fc667ee72dfe2b9073"
    client = serpapi.Client(api_key=API_KEY)
    
    # Setup search parameters
    params = {
        "engine": "google_flights",
        "departure_id": departure_id,
        "arrival_id": arrival_id,
        "outbound_date": outbound_date,
        "currency": currency,
        "hl": "en",
        "gl": "in"
    }
    
    # type: 1 = Round Trip, 2 = One Way
    if return_date:
        params["return_date"] = return_date
        params["type"] = "1"
    else:
        params["type"] = "2"

    print(f"🔍 Searching Google Flights: {departure_id} ➔ {arrival_id} on {outbound_date}...")
    
    try:
        results = client.search(params)
    except Exception as e:
        print(f"❌ SerpApi Request Failed: {e}")
        return []

    booking_url = results.get("search_metadata", {}).get("google_flights_url", "")
    best_flights = results.get("best_flights", [])
    
    if not best_flights:
        # Fallback to other_flights if best_flights is empty
        best_flights = results.get("other_flights", [])

    flight_options = []

    for item in best_flights[:5]:
        legs = item.get("flights", [])
        if not legs:
            continue
            
        first_leg = legs[0]   # True departure
        last_leg = legs[-1]   # Final arrival (even if connecting)
        
        # Combine unique airline names
        airlines = list(dict.fromkeys(leg.get("airline") for leg in legs if leg.get("airline")))
        
        flight_options.append({
            "airline": ", ".join(airlines),
            "flight_number": first_leg.get("flight_number"),
            "departure_airport": first_leg.get("departure_airport", {}).get("name"),
            "departure_time": first_leg.get("departure_airport", {}).get("time"),
            "arrival_airport": last_leg.get("arrival_airport", {}).get("name"),
            "arrival_time": last_leg.get("arrival_airport", {}).get("time"),
            "duration_minutes": item.get("total_duration"),
            "stops": len(legs) - 1,
            "price_inr": item.get("price"),
            "booking_url": booking_url
        })
        
    return flight_options


# ==========================================
# 🧪 RUN TEST
# ==========================================
test_results = search_google_flights(
    departure_id="HYD",
    arrival_id="DEL",
    outbound_date="2026-11-09"
)

# 1. Print Human Summary
print(f"\n✅ Total options found: {len(test_results)}\n")
for i, f in enumerate(test_results, 1):
    stop_text = "Direct" if f['stops'] == 0 else f"{f['stops']} Stop(s)"
    print(f"Option {i}: {f['airline']} ({f['flight_number']}) | {stop_text}")
    print(f"  🛫 Departure: {f['departure_time']} ({f['departure_airport']})")
    print(f"  🛬 Arrival:   {f['arrival_time']} ({f['arrival_airport']})")
    print(f"  ⏱ Duration:  {f['duration_minutes']} mins | 💰 Price: ₹{f['price_inr']}")
    print("-" * 50)

# 2. View the Structured JSON (This goes into your Shared State)
print("\n📦 Structured Data for Shared State:")
print(json.dumps(test_results[:1], indent=2))
```
